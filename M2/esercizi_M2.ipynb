### Consegna:

te, tre voti in virgola mobile e le ore di assenza, calcola la media dei tre voti e stampa il giudizio

```python
nomeS = "Chiara"
votiS = 7.5, 8, 6.5
assenzeS = 12
media = round(sum(votiS)/3, 2)
giudizio = " "

if assenzeS > 50:
    giudizio = "non valutabile"
elif media > 8:
    giudizio = "ottimo"
elif media > 6:
    giudizio = "sufficiente"
elif media < 6:
    giudizio = "insufficiente"

print(f"{nomeS}: media {media} {giudizio}")
```

# Esercizio 5 - Conteggio dei caratteri di una frase
### Consegna:
- Scrivere nella cella dell'esercizio 5 il programma che legge in ingresso una frase e conta, con
un unico ciclo for sui caratteri, quante vocali, quante consonanti, quante cifre e quanti altri
caratteri contiene. Il programma stampa i quattro conteggi e la percentuale di vocali sul totale
dei caratteri, con un decimale. Le lettere accentate vanno considerate consonanti: è una
semplificazione voluta, da dichiarare in un commento.

```python
frase = "Il laboratorio 4BI dura 22 ore"
vocali = 0
consonanti = 0
cifre = 0
altri = 0

for carattere in frase.lower():
    if carattere in "aeiou":
        vocali += 1
    elif carattere.isalpha():
        consonanti +=1
    elif carattere.isdigit():
        cifre += 1
    elif carattere:
        altri += 1
perfentualeV = (vocali/len(frase))*100

print(f"voacali: {vocali}, consonanti: {consonanti}, cifre: {cifre}, altri {altri}")
print(f"percentuale di vocali: {perfentualeV}%")
```

# Esercizio 7 - Analisi delle cifre di un numero
### Consegna:
- Scrivere nella cella dell'esercizio 7 il programma che legge in ingresso un intero non negativo
e ne calcola, con un ciclo while che usa soltanto il resto e la divisione intera per dieci, la
somma delle cifre, il numero di cifre e la cifra maggiore. Il numero zero va gestito correttamente: 
ha una cifra, la somma è zero e la cifra maggiore è zero. Non è ammesso trasformare il
numero in stringa.

```python
numero = 48723
somma = 0
cifra = 0
max = 0

if numero == 0: cifra = 1
while numero > 0:
    somma += numero % 10
    if numero%10>max:
        max = numero%10
    numero = numero // 10
    cifra += 1

print(f"cifre: {cifra}")
print(f"somme delle cifre: {somma}")
print(f"cifra maggiore: {max}")
```

# Esercizio 9 - Ricerca sequenziale in un elenco di codici
### Consegna:
- Scrivere nella cella dell'esercizio 9 il programma che, a partire dalla stringa
"A101;B204;C307;D410;E512" , legge in ingresso un codice e ne cerca la presenza con una
scansione voce per voce: il programma estrae le voci una alla volta con find e slicing e le
confronta con il codice cercato, interrompendo la scansione appena lo trova. Va stampata la
posizione della voce contando da uno, oppure il messaggio di codice assente. La ricerca è insensibile 
agli spazi agli estremi e alle maiuscole. Il caso «non trovato» va gestito con la clausola 
else del ciclo, senza variabili booleane di appoggio.

```python
stringa = "A101;B204;C307;D410;E512"
codice = "c307".strip().upper()

inizio = 0
posizione = 1

while inizio < len(stringa):
    fine = stringa.find(';', inizio)
    if fine == -1:
        voce = stringa[inizio:].strip().upper()
        inizio = len(stringa)
    else:
        voce = stringa[inizio:fine].strip().upper()
        inizio = fine + 1
    if voce == codice:
        print(f"{codice} è la voce numero {posizione} dell'elenco")
        break
    posizione += 1
else:
    print(f"{codice} non è presente nell'elenco")
```
