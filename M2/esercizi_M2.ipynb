# ESERCIZIO 1 - Durata di una sessione di laboratorio
### consegna: 
- Scrivere nella cella dell'esercizio 1 il programma che legge in ingresso la durata di una sessio‐  
ne di laboratorio espressa in secondi e la stampa nel formato ore, minuti e secondi, usando  
soltanto la divisione intera e il resto. Il programma stampa anche la stessa durata espressa in   
ore decimali con due cifre dopo la virgola, e segnala con un messaggio il caso in cui la durata  
superi le quattro ore.

```python
variabileNum = 9296

ore = variabileNum // 3600
minuti = (variabileNum // 60) % 60
secondi = variabileNum % 60

ore_decimali = round(float(variabileNum/3600), 2)

if variabileNum >= 14400:
    print("attenzione: la sessione supera le quattro ore")


print(f"{ore:02d}:{minuti:02d}:{secondi:02d}")
print(f"in ore decimali: {ore_decimali}")
```

# ESERCIZIO 2 - Media dei voti e giudizio
### Consegna:
- Scrivere nella cella dell'esercizio 2 il programma che legge in ingresso il nome di uno studen‐
te, tre voti in virgola mobile e le ore di assenza, calcola la media dei tre voti e stampa il giudizio
secondo questa regola: se le ore di assenza superano 50 il giudizio è non valutabile , indi‐
pendentemente dalla media; altrimenti il giudizio è ottimo da 8 in su, sufficiente da 6 in
su, insufficiente sotto il 6. La media va stampata con due decimali dentro una f-string,
insieme al nome dello studente.

```python
nomeS = "Chiara"
votiS = 7.5, 8, 6.5
assenzeS = 12
media = round(sum(votiS)/3, 2)
giudizio = " "

if assenzeS > 50:
    giudizio = "non valutabile"
elif media > 8:
    giudizio = "ottimo"
elif media > 6:
    giudizio = "sufficiente"
elif media < 6:
    giudizio = "insufficiente"

print(f"{nomeS}: media {media} {giudizio}")
```
